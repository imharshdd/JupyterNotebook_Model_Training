```python
import requests
from bs4 import BeautifulSoup
import csv
from apscheduler.schedulers.blocking import BlockingScheduler
import datetime

def scrape_option_chain():
    try:
        # URL of the NSE option chain website
        nse_option_chain_url = 'https://www.nseindia.com/option-chain'

        # Send a GET request to the URL
        response = requests.get(nse_option_chain_url, headers={'User-Agent': 'Mozilla/5.0'})

        # Check if the request was successful
        if response.status_code == 200:
            # Parse the HTML content
            soup = BeautifulSoup(response.text, 'html.parser')
            
            # Find the total open interest data
            total_open_interest = soup.find('div', {'id': 'totalOI'}).text.strip()
            
            # Get the current timestamp
            current_time = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

            # Write data to CSV file
            with open('option_chain_data.csv', 'a', newline='') as csvfile:
                writer = csv.writer(csvfile)
                writer.writerow([current_time, total_open_interest])
            
            print('Data stored in option_chain_data.csv')
            
        else:
            print('Failed to fetch option chain data:', response.status_code)
    except Exception as e:
        print('Error occurred while scraping:', e)

# Create a scheduler
scheduler = BlockingScheduler()

# Schedule the scraping function to run every 1 minute
scheduler.add_job(scrape_option_chain, 'interval', minutes=1)

try:
    # Start the scheduler
    scheduler.start()
except KeyboardInterrupt:
    # Stop the scheduler when KeyboardInterrupt (Ctrl+C) is received
    scheduler.shutdown()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from apscheduler.schedulers.blocking import BlockingScheduler
import datetime

def scrape_option_chain():
    try:
        # URL of the NSE option chain website
        nse_option_chain_url = 'https://www.nseindia.com/option-chain'

        # Send a GET request to the URL
        response = requests.get(nse_option_chain_url, headers={'User-Agent': 'Mozilla/5.0'})

        # Check if the request was successful
        if response.status_code == 200:
            # Parse the HTML content
            soup = BeautifulSoup(response.text, 'html.parser')
            
            # Find the total open interest data
            total_open_interest = soup.find('div', {'id': 'totalOI'}).text.strip()
            
            # Get the current timestamp
            current_time = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
            
            # Store data in DataFrame
            data = {'Timestamp': [current_time], 'Total Open Interest': [total_open_interest]}
            df = pd.DataFrame(data)
            
            # Write data to CSV file
            df.to_csv('option_chain_data.csv', mode='a', index=False, header=not df.index.isin([0]).any())
            
            print('Data stored in option_chain_data.csv')
            
        else:
            print('Failed to fetch option chain data:', response.status_code)
    except Exception as e:
        print('Error occurred while scraping:', e)

# Create a scheduler
scheduler = BlockingScheduler()

# Schedule the scraping function to run every 1 minute
scheduler.add_job(scrape_option_chain, 'interval', minutes=1)

try:
    # Start the scheduler
    scheduler.start()
except KeyboardInterrupt:
    # Stop the scheduler when KeyboardInterrupt (Ctrl+C) is received
    scheduler.shutdown()
```
